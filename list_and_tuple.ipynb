# LIST

```python
# collection of similar or different data type of elements
# Homogenous 
print("Homogenous type of lists")
collection: list = [2,3,4,5]
collection1: list = ["A","B","C"]
collection2: list = [1.0,2.0,3.0]
print(collection)
print(collection1)
print(collection2)
print("\n")
print("Non Homogenous type of list")
collection3: list = ["A",1,1.0,True]
print(collection3)
```

```python
# list indexing:used to acces element ,iteration ,insertion
# index strat from 0
# why index strat from zero
print(collection[2])
print(collection[-1])  # negative index return element from last of the list element strat with -1,-2
```

```python
# lenght of list 
# len function return lenght of list which is strat from 1
print(len(collection))
```

```python
# lists are changeable
collection[3] = [6,7,8]
print(f"List after adding another list {collection}")
```
