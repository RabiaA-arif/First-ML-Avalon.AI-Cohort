**Regex Rule**

**'.'**   matches any character except the newline

**^** match the start of string and in multiline pattren immediately after newline

**$** Matches the end of the string or just before the newline at the end of the string

'*' : Causes the resulting RE to match 0 or more repetitions of the preceding RE

'+' Causes the resulting RE to match 1 or more repetitions of the preceding RE

'?' Causes the resulting RE to match 0 or 1 repetitions of the preceding RE

{m} : Specifies that exactly m copies of the previous RE should be matched

{m,n} : Causes the resulting RE to match from m to n repetitions of the preceding RE, attempting to match as many repetitions as possible

```python
import re
text = "we all are in library"
pattren = r"library"
match = re.findall(pattren,text)
print(match)
```

```python
import re
text = "there is no internet in pakistan"
x = re.search("there is",text) # check is string start with 'there'
if x:
  print("string match")
else:
  print("string is not match")
# print(x)
```

```python
import re
txt = "tommorow is my paper"
find = re.findall("paer",txt)
print(find) # return empty list where no match found
if find:
  print("we find word")
else:
  print("This word is not exist in our txt")
```

```python
import re
txt = "I am from umerkot"
# txt1 = "iamfrom"
serch = re.search(r"\s",txt)
split = re.split(r"am",txt)
print(f"Splited string is:{split}")
print(f"first white space is start from index:{serch.start()}")
```

```python
# sub() replaces the matches  with given text
import re
txt = "i am Rabia"
replc = re.sub(r"Rabia","arif",txt)
print(f"Replaced string is:{replc}")
```
