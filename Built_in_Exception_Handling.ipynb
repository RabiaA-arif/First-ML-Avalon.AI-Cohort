## **<font color="green"> Control Flow Of Exception</font>**

**try:** in try block we write code in which error accur at runtime

**except:** give detail about error which happen in `try` block

**else:** run code when try block execute with any error

**finally:** Always executes at the very end, regardless of whether an error occurred or was handled, making it perfect for cleanup tasks like closing files

```python
def reraise(tp, value, tb=None):
    if value is None:
        value = tp()
    if value.__traceback__ is not tb:
        raise value.with_traceback(tb)
    raise value
```

```python
def catch():
    try:
        result = bar()
    except Exception as e:
        result = e
    try:
        return result
    finally:
        del result
def reraise(tp, value, tb=None):
    if value is None:
        value = tp()
    try:
        if value.__traceback__ is not tb:
            raise value.with_traceback(tb)
        raise value
    finally:
        del value, tb
```
