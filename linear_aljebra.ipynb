```python
# sir's code
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from matplotlib.patches import Arc, Circle
import seaborn as sns


# 1. Cosine Similarity Function
# ==========================================
def cosine_similarity(a, b):
    a = np.array(a, dtype=float)
    b = np.array(b, dtype=float)
    dot = np.dot(a, b)
    norm_a = np.linalg.norm(a)
    norm_b = np.linalg.norm(b)
    if norm_a == 0 or norm_b == 0:
        return 0.0
    return dot / (norm_a * norm_b)


# ==========================================
# 2. 2D Vectors + Angle + Unit Circle
# ==========================================
def plot_2d_with_unit_circle(a, b, title="2D Vectors + Unit Circle"):
    a = np.array(a, dtype=float)[:2]
    b = np.array(b, dtype=float)[:2]

    # Normalize for unit circle
    a_unit = a / np.linalg.norm(a)
    b_unit = b / np.linalg.norm(b)

    sim = cosine_similarity(a, b)
    angle_deg = np.degrees(np.arccos(np.clip(sim, -1.0, 1.0)))

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

    # ----- Left: Original Vectors -----
    origin = np.array([0, 0])
    ax1.quiver(*origin, *a, color='blue', scale_units='xy', scale=1, width=0.015, label=f'A {a}')
    ax1.quiver(*origin, *b, color='red',  scale_units='xy', scale=1, width=0.015, label=f'B {b}')

    # Angle arc
    angle_a = np.degrees(np.arctan2(a[1], a[0]))
    angle_b = np.degrees(np.arctan2(b[1], b[0]))
    theta1, theta2 = sorted([angle_a, angle_b])
    if theta2 - theta1 > 180:
        theta1, theta2 = theta2, theta1 + 360

    radius = max(np.linalg.norm(a), np.linalg.norm(b)) * 0.3
    arc = Arc(origin, radius*2, radius*2, angle=0, theta1=theta1, theta2=theta2, color='green', lw=2)
    ax1.add_patch(arc)

    ax1.set_xlim(-max(np.linalg.norm(a), np.linalg.norm(b))*1.4,
                 max(np.linalg.norm(a), np.linalg.norm(b))*1.4)
    ax1.set_ylim(-max(np.linalg.norm(a), np.linalg.norm(b))*1.4,
                 max(np.linalg.norm(a), np.linalg.norm(b))*1.4)
    ax1.set_aspect('equal')
    ax1.grid(True, linestyle='--', alpha=0.6)
    ax1.axhline(0, color='black', lw=0.8)
    ax1.axvline(0, color='black', lw=0.8)
    ax1.set_title("Original Vectors")
    ax1.legend()
    ax1.text(0.05, 0.95, f"Cosine = {sim:.4f}\nAngle ≈ {angle_deg:.1f}°",
             transform=ax1.transAxes, fontsize=11, verticalalignment='top',
             bbox=dict(boxstyle='round', facecolor='wheat'))

    # ----- Right: Unit Circle (Normalized) -----
    circle = Circle((0, 0), 1, fill=False, color='gray', linestyle='--', lw=1.5)
    ax2.add_patch(circle)

    ax2.quiver(*origin, *a_unit, color='blue', scale_units='xy', scale=1, width=0.015, label='A (normalized)')
    ax2.quiver(*origin, *b_unit, color='red',  scale_units='xy', scale=1, width=0.015, label='B (normalized)')

    # Angle arc on unit circle
    arc2 = Arc(origin, 0.6, 0.6, angle=0, theta1=theta1, theta2=theta2, color='green', lw=2)
    ax2.add_patch(arc2)

    ax2.set_xlim(-1.5, 1.5)
    ax2.set_ylim(-1.5, 1.5)
    ax2.set_aspect('equal')
    ax2.grid(True, linestyle='--', alpha=0.6)
    ax2.axhline(0, color='black', lw=0.8)
    ax2.axvline(0, color='black', lw=0.8)
    ax2.set_title("Unit Circle (Normalized Vectors)")
    ax2.legend()
    ax2.text(0.05, 0.95, "Length = 1\nSirf direction matter karti hai",
             transform=ax2.transAxes, fontsize=11, verticalalignment='top',
             bbox=dict(boxstyle='round', facecolor='lightblue'))

    plt.suptitle(title, fontsize=14, fontweight='bold')
    plt.tight_layout()
    plt.show()


# ==========================================
# 3. Multiple Vectors → Heatmap
# ==========================================
def plot_similarity_heatmap(vectors, labels=None):
    vectors = np.array(vectors, dtype=float)
    n = len(vectors)

    # Similarity matrix
    sim_matrix = np.zeros((n, n))
    for i in range(n):
        for j in range(n):
            sim_matrix[i, j] = cosine_similarity(vectors[i], vectors[j])

    if labels is None:
        labels = [f"V{i+1}" for i in range(n)]

    plt.figure(figsize=(8, 6))
    sns.heatmap(sim_matrix, annot=True, fmt=".3f", cmap="YlGnBu",
                xticklabels=labels, yticklabels=labels,
                vmin=-1, vmax=1, square=True, linewidths=0.5)
    plt.title("Cosine Similarity Heatmap", fontsize=14, fontweight='bold')
    plt.tight_layout()
    plt.show()

    return sim_matrix


# ==========================================
# 4. 3D Vectors Visualization
# ==========================================
def plot_3d_vectors(a, b, title="3D Vectors"):
    a = np.array(a, dtype=float)[:3]
    b = np.array(b, dtype=float)[:3]

    sim = cosine_similarity(a, b)

    fig = plt.figure(figsize=(9, 7))
    ax = fig.add_subplot(111, projection='3d')

    origin = [0, 0, 0]

    ax.quiver(*origin, *a, color='blue', arrow_length_ratio=0.1, linewidth=2, label=f'A {a}')
    ax.quiver(*origin, *b, color='red',  arrow_length_ratio=0.1, linewidth=2, label=f'B {b}')

    # Limits
    max_range = max(np.linalg.norm(a), np.linalg.norm(b)) * 1.3
    ax.set_xlim([-max_range, max_range])
    ax.set_ylim([-max_range, max_range])
    ax.set_zlim([-max_range, max_range])

    ax.set_xlabel('X')
    ax.set_ylabel('Y')
    ax.set_zlabel('Z')
    ax.set_title(f"{title}\nCosine Similarity = {sim:.4f}", fontsize=13)
    ax.legend()
    plt.tight_layout()
    plt.show()


# ======================================================
# Running All Examples
# ======================================================

print(">>> Example 1: 2D + Unit Circle (High Similarity)")
plot_2d_with_unit_circle([3, 4], [4, 5], "High Similarity Example")

print("\n>>> Example 2: Orthogonal Vectors")
plot_2d_with_unit_circle([5, 0], [0, 4], "Orthogonal (90°)")

print("\n>>> Example 3: Opposite Direction")
plot_2d_with_unit_circle([3, 2], [-3, -2], "Opposite Direction")

print("\n>>> Example 4: Multiple Vectors Heatmap")
vectors = [
    [1, 2, 3],
    [2, 3, 4],
    [1, 0, 0],
    [0, 1, 0],
    [3, 1, 2],
    [-1, -2, -3]
]
labels = ["Doc1", "Doc2", "X-axis", "Y-axis", "Doc3", "Opposite"]
sim_matrix = plot_similarity_heatmap(vectors, labels)

print("\n>>> Example 5: 3D Vectors")
plot_3d_vectors([2, 3, 4], [3, 1, 5], "3D Vectors Example")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from matplotlib.patches import FancyBboxPatch, Circle, Arc
from matplotlib.collections import LineCollection
import seaborn as sns
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.feature_extraction.text import TfidfVectorizer
import warnings
warnings.filterwarnings("ignore")

# ====================== THEME ======================
plt.style.use('dark_background')
plt.rcParams['figure.facecolor'] = '#0b0e14'
plt.rcParams['axes.facecolor']   = '#0b0e14'
plt.rcParams['savefig.facecolor'] = '#0b0e14'
plt.rcParams['font.family'] = 'DejaVu Sans'
plt.rcParams['axes.titleweight'] = 'bold'

# ====================== SAMPLE RAG DATA ======================
documents = [
    "Retrieval Augmented Generation combines search with large language models",
    "Cosine similarity is widely used to rank relevant documents in RAG systems",
    "Vector databases store embeddings for fast semantic search",
    "Transformer models generate powerful text embeddings for retrieval",
    "Traditional keyword search fails to capture semantic meaning of queries",
    "FAISS and Pinecone are popular vector databases used in production RAG",
    "The weather is sunny today and perfect for outdoor activities",
    "Cooking pasta requires boiling water and adding salt"
]

query = "How does cosine similarity help in retrieval augmented generation?"

# ====================== TF-IDF + COSINE ======================
vectorizer = TfidfVectorizer(stop_words='english')
doc_vectors = vectorizer.fit_transform(documents)
query_vector = vectorizer.transform([query])

# Cosine similarity scores
scores = cosine_similarity(query_vector, doc_vectors).flatten()
ranked_indices = np.argsort(scores)[::-1]

print("\nRAG Retrieval Ranking (Cosine Similarity):")
print("="*70)
for rank, idx in enumerate(ranked_indices, 1):
    print(f"{rank:2d}. [{scores[idx]:.4f}]  {documents[idx]}")

# ====================== 1. ANIMATED RANKING BAR CHART ======================
def animate_rag_ranking():
    fig, ax = plt.subplots(figsize=(12, 7))

    colors = plt.cm.plasma(np.linspace(0.2, 0.9, len(documents)))
    bars = ax.barh(range(len(documents)), np.zeros(len(documents)), color=colors, height=0.7)

    ax.set_yticks(range(len(documents)))
    ax.set_yticklabels([f"Doc {i+1}" for i in range(len(documents))], fontsize=11)
    ax.set_xlim(0, 1.05)
    ax.set_xlabel("Cosine Similarity Score", fontsize=13, color='#e0e0e0')
    ax.set_title("RAG Document Ranking — Cosine Similarity Animation",
                 fontsize=16, pad=20, color='#00ffcc')

    # Glowing query box
    query_box = FancyBboxPatch((0.02, 0.92), 0.96, 0.06, transform=ax.transAxes,
                               boxstyle="round,pad=0.02", facecolor='#1a1a2e',
                               edgecolor='#00ffcc', linewidth=2, alpha=0.9)
    ax.add_patch(query_box)
    ax.text(0.5, 0.95, f"Query: {query}", transform=ax.transAxes,
            ha='center', va='center', fontsize=11, color='#00ffcc')

    def update(frame):
        current_scores = scores * (frame / 40)
        for bar, score, color in zip(bars, current_scores[ranked_indices], colors):
            bar.set_width(score)
            bar.set_color(color)

        # Highlight top-3
        if frame > 30:
            for i in range(min(3, len(bars))):
                bars[i].set_edgecolor('#00ffcc')
                bars[i].set_linewidth(2.5)

        return bars

    ani = animation.FuncAnimation(fig, update, frames=45, interval=60, blit=False, repeat=False)
    plt.tight_layout()
    plt.show()
    return ani

# ====================== 2. BEAUTIFUL SIMILARITY HEATMAP ======================
def plot_beautiful_heatmap():
    sim_matrix = cosine_similarity(doc_vectors)

    fig, ax = plt.subplots(figsize=(11, 9))

    mask = np.triu(np.ones_like(sim_matrix, dtype=bool), k=1)

    sns.heatmap(sim_matrix,
                mask=mask,
                annot=True,
                fmt=".2f",
                cmap="magma",
                center=0.4,
                square=True,
                linewidths=1.5,
                linecolor='#1a1a2e',
                cbar_kws={'label': 'Cosine Similarity', 'shrink': 0.8},
                ax=ax)

    ax.set_title("Document-to-Document Cosine Similarity Matrix\n(RAG Knowledge Base)",
                 fontsize=15, color='#00ffcc', pad=15)

    labels = [f"D{i+1}" for i in range(len(documents))]
    ax.set_xticklabels(labels, rotation=0, color='#e0e0e0')
    ax.set_yticklabels(labels, rotation=0, color='#e0e0e0')

    plt.tight_layout()
    plt.show()

# ====================== 3. ANIMATED VECTOR SPACE (2D PCA style) ======================
def animate_vector_space():
    # Reduce to 2D for visualization (simple projection)
    from sklearn.decomposition import TruncatedSVD
    svd = TruncatedSVD(n_components=2, random_state=42)
    doc_2d = svd.fit_transform(doc_vectors)
    query_2d = svd.transform(query_vector)[0]

    fig, ax = plt.subplots(figsize=(11, 9))

    # Background glow
    ax.set_facecolor('#0b0e14')

    scatter = ax.scatter(doc_2d[:, 0], doc_2d[:, 1],
                         c=scores, cmap='plasma', s=180,
                         edgecolors='white', linewidth=1.5, zorder=5)

    # Query point
    query_point = ax.scatter(query_2d[0], query_2d[1],
                             c='#00ffcc', s=350, marker='*',
                             edgecolors='white', linewidth=2, zorder=10, label='Query')

    # Animated lines from query to documents
    lines = []
    for i in range(len(documents)):
        line, = ax.plot([], [], lw=1.5, alpha=0.6, color='#00ffcc')
        lines.append(line)

    # Annotations
    for i, (x, y) in enumerate(doc_2d):
        ax.annotate(f"D{i+1}", (x, y), textcoords="offset points",
                    xytext=(8, 8), fontsize=9, color='white', alpha=0.8)

    ax.set_title("RAG Vector Space — Query vs Documents\n(Animated Cosine Connections)",
                 fontsize=15, color='#00ffcc', pad=15)
    ax.legend(loc='upper right', fontsize=11)
    ax.grid(True, alpha=0.15, color='white')

    def update(frame):
        progress = min(frame / 30, 1.0)
        for i, line in enumerate(lines):
            x = [query_2d[0], query_2d[0] + (doc_2d[i, 0] - query_2d[0]) * progress]
            y = [query_2d[1], query_2d[1] + (doc_2d[i, 1] - query_2d[1]) * progress]
            line.set_data(x, y)
            # Thickness based on similarity
            line.set_linewidth(0.5 + scores[i] * 4)
            line.set_alpha(0.3 + scores[i] * 0.7)
        return lines

    ani = animation.FuncAnimation(fig, update, frames=45, interval=50, blit=True, repeat=False)
    plt.tight_layout()
    plt.show()
    return ani

# ====================== 4. UNIT CIRCLE + ANGLE ANIMATION ======================
def animate_unit_circle():
    # Take top document and query (simplified 2D)
    top_idx = ranked_indices[0]

    # Create two unit vectors with realistic angle
    angle = np.arccos(np.clip(scores[top_idx], -1, 1))

    fig, ax = plt.subplots(figsize=(9, 9))
    ax.set_xlim(-1.4, 1.4)
    ax.set_ylim(-1.4, 1.4)
    ax.set_aspect('equal')

    # Unit circle
    circle = Circle((0, 0), 1, fill=False, color='#00ffcc', lw=2, alpha=0.7)
    ax.add_patch(circle)

    # Vectors
    query_vec, = ax.plot([], [], color='#00ffcc', lw=3, label='Query (normalized)')
    doc_vec,   = ax.plot([], [], color='#ff6b6b', lw=3, label='Top Document')

    # Angle arc
    arc_line, = ax.plot([], [], color='#ffd93d', lw=2.5)

    ax.set_title("Cosine Similarity on Unit Circle\n(RAG Top Retrieved Document)",
                 fontsize=15, color='#00ffcc')
    ax.legend(loc='upper left')
    ax.axhline(0, color='white', alpha=0.2)
    ax.axvline(0, color='white', alpha=0.2)
    ax.grid(True, alpha=0.15)

    def update(frame):
        t = min(frame / 40, 1.0)
        current_angle = angle * t

        # Query always on x-axis
        qx, qy = 1.0, 0.0
        dx = np.cos(current_angle)
        dy = np.sin(current_angle)

        query_vec.set_data([0, qx], [0, qy])
        doc_vec.set_data([0, dx], [0, dy])

        # Arc
        theta = np.linspace(0, current_angle, 50)
        arc_line.set_data(0.35 * np.cos(theta), 0.35 * np.sin(theta))

        return query_vec, doc_vec, arc_line

    ani = animation.FuncAnimation(fig, update, frames=50, interval=40, blit=True, repeat=False)
    plt.tight_layout()
    plt.show()
    return ani

# ====================== RUN ALL ======================
print("\nLaunching High-Class RAG Visualizations...\n")

ani1 = animate_rag_ranking()
plot_beautiful_heatmap()
ani2 = animate_vector_space()
ani3 = animate_unit_circle()

print("\nAll animations completed successfully.")
```
