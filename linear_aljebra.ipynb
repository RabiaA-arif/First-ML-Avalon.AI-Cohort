```python
# sir babar code
import numpy as np

A = np.array([[3,2],[1,3]])
v1 = np.array([4, 2])
v2 = np.array([3, -3])
c=5

print("T(c*v1)  =", A @ (c * v1))
print("c*T(v1)  =", c * (A @ v1))
print("Equal?   =", np.allclose(A @ (c*v1), c*(A @ v1)))

```

```python
# sir babar code
import matplotlib.pyplot as plt

# Ek grid of points banate hain

x = np.linspace(-2,3,10)

y = np.linspace(2,3,3)

X,Y = np.meshgrid(x,y)

points = np.vstack([X.ravel(), Y.ravel()])

A = np.array([[3,1], [3,1]])

transformed = A @ points

fig,axs= plt.subplots(1,2, figsize= (10,5))

axs[0].scatter(points[0], points[1], c = "steelblue")

axs[0].set_title("Before (Original Space)")

axs[1].scatter(transformed[0], transformed[1], c = "crimson")

axs[1].set_title("After (Transformed Space)")

for ax in axs:

  ax.axhline(0, color = "gray", lw=0.4)

  ax.axhline(0, color = "gray", lw=0.4)

  ax.set_aspect("equal")

plt.show()
```
