## Build a calculator

```python
!python --version
import math
```

```python
# simplest version of calculator
class Calculator:
  def __init__(self, x1, x2):
    self.x1 = x1
    self.x2 = x2

  def add(self):
    return self.x1 + self.x1

  def subtraction(self):
    return self.x1 - self.x2

  def multiplication(self):
    return self.x1 * self.x2


```

```python
calculator = Calculator(22,3)
print(calculator.multiplication())
```

```python
# Intermidiate level
# how to handle if multiple values comes for mathematical operation
class Calculator1:
  # def __init__(self, *args):
  #   self.args = args
  def add(*args):
    return f"addition is :{sum(args)}"

  def multiplication(*args):
    return f"Multiplication:{math.prod(args)}"

  def divide(num1, num2):
    return f"Division:{num1 / num2}"






```

```python
ob = Calculator1
ob.add(1,2,3,4)
ob.multiplication(22,33,44)
ob.divide(num2=22, num1=33)
```

## Count Word frequency

```python
count_words = 0
sentence = "Fashion trends are changing like day and night,why trend changing ,can you explain the fashion trends"
words = sentence.split()
words_counts = {}
for word in words:
  words_counts[word] = words_counts.get(word,0) + 1
print(words_counts)
```

## Find duplicate in list

```python
numbers = [1,2,3,2,3,4,5]
not_dublicate = []
dublicate = []
for i in numbers:
  if i not in not_dublicate:
    not_dublicate.append(i)
  else:
    dublicate.append(i)

print(not_dublicate)
print(dublicate)
```

### Create a Student class with attributes such as name, age, marks, and methods to calculate the average and display student information.

```python
class Student:
  def __init__(self, name, age, marks):
    self.name = name
    self.age = age
    self.marks = marks

  def average_marks(self):
    return f"Average marks is:{sum(self.marks)/len(self.marks)}"
  def student_information(self):
    return {

        "Name":self.name,
        "Age":self.age,
        "Marks":self.marks
    }
```

```python
student1 = Student('Azlan', 19,[1,2,3,4])
student1.average_marks()
student1.student_information()
```
