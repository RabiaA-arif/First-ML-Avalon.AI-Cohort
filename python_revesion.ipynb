## Build a calculator

```python
# simplest version of calculator
class Calculator:
  def __init__(self, x1, x2):
    self.x1 = x1
    self.x2 = x2

  def add(self):
    return self.x1 + self.x1

  def subtraction(self):
    return self.x1 - self.x2

  def multiplication(self):
    return self.x1 * self.x2


```

```python
calculator = Calculator(22,3)
print(calculator.multiplication())
```

```python
# Intermidiate level
# how to handle if multiple values comes for mathematical operation
class Calculator1:
  # def __init__(self, *args):
  #   self.args = args
  def add(*args):
    return f"addition is :{sum(args)}"

  def multiplication(*args):
    return f"Multiplication:{math.prod(args)}"

  def divide(num1, num2):
    return f"Division:{num1 / num2}"






```

```python
ob = Calculator1
ob.add(1,2,3,4)
ob.multiplication(22,33,44)
ob.divide(num2=22, num1=33)
```

## Find duplicate in list

```python
numbers = [1,2,3,2,3,4,5]
not_dublicate = []
dublicate = []
for i in numbers:
  if i not in not_dublicate:
    not_dublicate.append(i)
  else:
    dublicate.append(i)

print(not_dublicate)
print(dublicate)
```

### Create a Student class with attributes such as name, age, marks, and methods to calculate the average and display student information.

```python
class Student:
  def __init__(self, name, age, marks):
    self.name = name
    self.age = age
    self.marks = marks

  def average_marks(self):
    return f"Average marks is:{sum(self.marks)/len(self.marks)}"
  def student_information(self):
    return {

        "Name":self.name,
        "Age":self.age,
        "Marks":self.marks
    }
```

Create a BankAccount class where the balance cannot be changed directly and withdrawals cannot exceed the available balance

```python
class BankAccountz:
    def __init__(self, balance):
        self.__balance = balance
        
    
    @property
    def balance(self):
        return self.__balance
    
    
    def withdraw_money(self, amount):
        if amount > self.__balance:
            print("Transaction Error ,Your amount  is more the than total balance")
            
        else:
            self.__balance -= amount
            print(f"Your amount is successfully withdraw which is : {amount}")
```

```python
acc = BankAccountz(1200)
acc.withdraw_money(100)
```

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    @property
    def balance(self):
        return self.__balance

    def withdraw(self, amount):
        if amount > self.__balance:
            print("Withdrawal exceeds the available balance.")
        else:
            self.__balance -= amount


acc = BankAccount(200)
acc.withdraw(130)
print(acc.balance)  # 70
```

Create a base Employee class and child classes such as Developer, Manager, and Designer. Override a method such as calculate_salary() or display_role().

single inheritance with overloading

```python
class Employee:
    def __init__(self, name, salary, role):
        self.name = name
        self.salary = salary
        self.role = role
    
    def calculate_salary(self):
        return self.salary
    
    def display_role(self):
        return self.display_role
    
class Developer(Employee):
    def calculate_salary(self):
        return super().calculate_salary()
    
    def display_role(self):
        return super().display_role()
class Manager(Employee):
    def calculate_salary(self):
        return super().calculate_salary()
    def display_role(self):
        return super().display_role()
class Designer(Employee):
    def calculate_salary(self):
        return super().calculate_salary()
    
    def display_role(self):
        return super().display_role()
```

```python
emp = Manager("Taha", 100000, 'manager')
emp.calculate_salary()
```

super() and method overriding

```python
class Library():
    def __init__(self,librarian_count, sweaper):
        self.librarian_count = librarian_count
        self.sweaper = sweaper
        
    def show(self):
        print(f"Library staff ,Librarian:{self.librarian_count} and Sweaper:{self.sweaper}")
        
        
        
class Hostel(Library):
    def __init__(self,librarian_count, sweaper, wardens):
        super().__init__(librarian_count, sweaper)
        self.wardens = wardens
        
    
    def show(self):
        print(f"Library staff ,Librarian:{self.librarian_count} and Sweaper:{self.sweaper} and wardens:{self.wardens}")
```

```python
hostel = Hostel(2,2,3)
print(hostel.show())
```

Polymorphisim

```python
class Farmer:
    def __init__(self, name, land):
        self.name = name 
        self.land = land
    
    def info(self):
        return f"Name:{self.name} and Land:{self.land}"
    
class Owner0(Farmer):
    def info(self):
        return "Owner give authority to Farmer"
    
class Owner1(Farmer):
    def info(self):
        return " Owner have bad behaviour towards farmer"
    
```

```python
def farmer_info(farmer):
    print(farmer.info())

farm = Owner0("ali",3)
farm1 = Owner1("aliyar",5)

print(farmer_info(farm))
print(farmer_info(farm1))
```

operator overloading

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    def __add__(self, other):
        return Point(self.x + other.x,self.y + other.y)
    
    def __str__(self):
        return f"Point({self.x},{self.y})"
    
    
    
```

```python
p1 = Point(1,3)
p2 = Point(3,5)

p3 = p1 + p2
print(p3)
```

```python
class Duck:
    def quack(self):
        return self.__repr__() + ": Quack"

class Person:
    def quack(self):
        return self.__repr__() + ": Person can Quacklike a Duck"
```
