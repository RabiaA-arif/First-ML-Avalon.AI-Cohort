Bank System

```python
class Banking:
  def __init__(self,balance,amount,pin,account_holder):
    self.__balance = balance
    self.amount = amount
    self.__pin = pin
    self.account_holder = account_holder

 # create account
 # atm card application
 # checkbook application
 # sign in into digital app
 # notify process
 # account history checking
 # transfering budgets
 # hanlde case if some notes is not available

  def deposit(self,amount):
    try:
      self.__balance += amount
      return f"Your {amount} deposited in {self.account_holder} account"
    except Exception as e:
      print(f"An Error Accured:{e}")
      print("Amount deposited is not successfull")
    finally:
      print("Deposit procces is stop!")

  def check_balance(self):
    return f"Your Total Balance is:{self.__balance}"

  def withdraw(self,amount):
    try:
      if amount <= 0:
        print("Enter valid amount")
      if amount > 0 and amount <= self.__balance:
        self.__balance -= amount
        print("Collect Your cash!")
    except:
      print("An Error Accured in Withdrawl process!")

  def withdraw_using_atm(self):
    print("Wellcome to ATM")
    print("Insert Your Depbit/Credit/Master card ")
    user_pin = int(input("Enter your ATM card pin:"))
    if  user_pin == self.__pin:
      user_amount = int(input("Enter your withdrawl amount:"))
      self.withdraw(user_amount)
    else:
      print("Invalid pin!")

objec = Banking(10000,3000,1234,"Rabia")
# objec.withdraw_using_atm()
objec.deposit(2000)
```
