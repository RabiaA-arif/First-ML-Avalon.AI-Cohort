**Instantiation**:creating concrete(real thing) object from the class


```python
class Area:
  def __init__(self,length):   # special method for object initilizer
    self.length = length
  def calculate_area(self):
    return self.length * self.length
```

```python
class Area:
  def __init__(self,leght):
    self.leght = leght
  def calculate_are(self):
    return self.leght * self.leght
```

```python
area_1 = Area(4)   # instantiation
# Accessing atrributes and methods
area_1.calculate_are()  # methods
print(area_1.leght)     # attributes
area_1.leght = 21
print(area_1.leght)
area_1.calculate_are()
```

**Name Mangling**

Double leading underscores lunch a behaviors in python known as name mangling

Name mangling is useful when for preventing name inheritance clashes

```python
class Information:
  def __init__(self,value):
    self.__value = value
  def __name(self):
    print(self.__value)


info_1 = Information("Welcome")
vars(info_1)

vars(Information)  # vars :return object dictionary


```

**Use of class**

Python classes can help you write more organized, structured, maintainable, reusable, flexible, and user friendly code

**Avoid use of class**

- Store only data. Use a data class, enumeration, or a named tuple instead.

- Provide a single method. Use a function instead.
- Additionally, you should avoid creating custom classes to wrap up functionality that’s available through built-in types or third-party classes. Use the type or third-party class directly.
