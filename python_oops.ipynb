```python
class Cart:
  def __init__(self):
    self.items = []

  @property
  def total(self):
    return sum(item["price"] * item["quantity"] for item in self.items)



cart = Cart()
cart.items.append({"name":"mouse","price":700,"quantity":2})
cart.items.append({"name":"usb","price":500,"quantity":1})

print(f"Total Cart Price:{cart.total}")

""" When someone writes cart.total, Python actually runs the total method behind the scenes and returns its result."""
```

```python
# string Representation
# .__str__  informal string representation
#.__repr__  formal string representation

class Car:
  def __init__(self,make,modal,year,color):
    self.make = make
    self.modal = modal
    self.year = year
    self.color = color
    self.start = False


  def __str__(self):
    return(
        self.make,
        self.modal,
        self.year,
        self.color

    )

  def __repr__(self):
    return(
        self.make,
        self.modal,
        self.year,
        self.color
    )

car = Car("bmw","2020","2020","red")

print(f"__str__ method output:{car.__str__()}")

print(f"__repr__ method output:{car.__repr__()}")


```

### class method with `@classmethod`


A class method is a method that takes the class object as its first argument instead of taking self.

In this case, the argument should be called cls, which is also a strong convention in Python

```python
class ThreeDpoint:
  def __init__(self,x,y,z):
    self.x = x
    self.y = y
    self.z = z

  def __iter__(self):
    yield from (self.x, self.y, self.z)

  @classmethod                  # staticmethod
  def from_sequence(cls,sequence):
    return cls*(sequence)

  def __repr__(self):
    return f"{type(self).__name__}({self.x},{self.y},{self.z})"

points = ThreeDpoint(2,4,5)
points.__repr__()
```

### Static Method with `@staticmethod`

These methods don’t take the instance or the class as an argument. So, they’re regular functions defined within a class. You could’ve also defined them outside the class as a stand-alone function.

```python
class ThreeDPoints:
  def __init__(self,a,b,c):
    self.a = a
    self.b = b
    self.c = c


  @classmethod
  def all_points(cls,points):
    return cls*(points)


  def __repr__(self):
    return f"{type(self).__name__},({self.a},{self.b},{self.c})"



  @staticmethod
  def wel_message(name):
    print(f"Hi {name} this is your threedpoints class")


point = ThreeDPoints(5,6,7)
point.wel_message("Rabia")
point.__repr__()
```
