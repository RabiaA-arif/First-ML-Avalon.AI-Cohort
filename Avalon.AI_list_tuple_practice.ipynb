```python
# Question 1
my_list: list = ["Aimen",20,3.67,True]
for i in range(len(my_list)):
    print(type(my_list[i]))
    
print("\n")
    
list1: list = ["rice",9.0,False,2]
for a in range(len(list1)):
    print(type(list1[a]))
print("\n")
list2: list = [True,"city",20,999,99]
for b in range(len(list2)):
    print(f"data types of  list elements {type(list2[b])}")
print("\n") 
list3: list = ["AI",3,2.0]
for c in range(len(list3)):
    print(f"data type{type(list3[c])}")
print("\n")
list4: list = [3333,912.00]
for d in range(len(list4)):
    print(f"Data type{type(list4[d])}")
print("\n")  
list5: list = ["coginition",True]
for e in range(len(list5)):
    print(f"data type {type(list5[e])}")
```
