# Section 5 : Mutable vs immutable warefare

```python
# Question 12
a = [1,2,3]
b = a
b[0] = [99] # [99,1,2,3]
a[0]  # [99]

id(a) == id(b)  # True ,same memory idress becuase we point b to a
```

```python
# muscle memory :reference vs copy drill
# [:] shallow copy 
# when we assign element to d list c is remain unchanged
c = [1,2,3,4]
d =c[:]
d[0] = [99]
print(d)
# c # unchanged
e = list(c)
e[0]=[99]
print(e) # chnaged
print(c) # still unchanged

```

```python
import copy

list_org = [1,2,3,4]
second_list = copy.copy(list_org)
print(second_list)
second_list[0]=[22]
print(second_list,"new list")
print(list_org,"original list")
third_list = copy.deepcopy(list_org)
print(third_list,"new list")
third_list[0] = [99]
print(third_list,"new list")
print(list_org,"original list")
```

```python
# create list from 5 different method
# [:]
# list()
# copy.copy()
# copy.deepcopy()
a = [10,20,30,40]
b = a[:]
b[0]=[50]
print(f"Original list : {a}")
print(f"List after changing:{b}")

# list ()
b = list(a)
b[0]=60
print(f"Original list : {a}")
print(f"List after changing:{b}")

# copy.copy
b = copy.copy(a)
b[0]=70
print(f"Original list : {a}")
print(f"List after changing:{b}")

# copy.deepcopy

b = copy.deepcopy(a)
b[0]=80
print(f"Original list : {a}")
print(f"List after changing:{b}")
```
