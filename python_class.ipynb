## Duck Typing is form of polymorphisim where types and class object is determined by it's behavior rather than it's explicitly inheritance from a particular class

**Explicit Inheritance** intensional and visible relationship between child/subclass and parent/superclass


```python
from os import name
class Person:
  def name(self):
    print(f"Name Of Person is : Rabia")



class Student:
  def name(self):
    print(f"We Are The Student Of Uni")



def person_info(info):
  info.name()


obj=Person()
obj1=Student()
print(person_info(obj))
print(person_info(obj1))
```
