# Why is String Interning Useful
## Memory Efficiency : by storing one '(copy of identical)' string can reduce the memory usage

## Performence Improvement : Interned strings allow for faster comparasion



```python
# Automatic interning identifiers

a : str ="Rabia"
b : str ="Rabia"
print(a is b) # in this 'is' check the address
#  'is' : identity comparasion : it check the both two veriable point the same memory or not
#  '==' check value not address

list_a :list =[1,1,1]
list_b : list =[1,1,1]
print(list_a is list_b) # False because they both point seperate memory address
print(list_a == list_b) # True : becaues they both have same content
list_c : list  = list_a
print(list_c is list_a) # True : because point same memory address
```

```python
# Automatic Interning of Short Strings

a : str = "RabiaArif"
b : str ="RabiaArif"
print(a is b) #True : short string automatically interned
```

```python
# long string
sent : str = "this_is_place_heaven_and_with_full_of_evilness_yeah"
sent1 : str = "this_is_place_heaven_and_with_full_of_evilness_yeah"
print(sent is sent1)

m = 'longer_string_that_is_not_interned'
n = 'longer_string_that_is_not_interned'
print(m is n)
```

```python
# Manual String
import sys
name : str = "Rabia Arif"
name2 : str = "Rabia Arif"
print(name is name2) #False

a :str =sys.intern("Rabia Arif")
b : str = sys.intern("Rabia Arif")
print(a is b) # True
```

```python
# Memory Savings with Interning
str_list = [sys.intern('repeat') for _ in range(1000)]
print(str_list[0] is str_list[999])
```

## What is string interning in Python? Which strings are interned automatically and which are not? How would you force interning, and when might that matter in production systems?

Python interns compile-time string constants and strings that look like valid Python identifiers (only letters, digits, underscores). Runtime-generated strings like user input are not interned. Use `sys.intern()` to force it. Useful for memory savings when storing millions of repeated strings.
Identifiers and string literals are interned automatically
Strings with spaces, punctuation, or created at runtime are not
`sys.intern('mystring')` forces interning — returns the canonical object
Useful in NLP pipelines or large dictionaries with repeated keys
