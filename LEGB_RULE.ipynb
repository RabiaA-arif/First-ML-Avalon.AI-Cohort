


*   LEGB is the scope levels of variables
*   LEGB define the oder in which python interpreter  check the variable scope level

###***Local Scope***
variable name which we define inside the function which is only accessible inside the function

###***Enclosing Scope***
also known as `non-local space` which only exists if we use nested functions , refer the scope of outer function that contains inner function
`When a variable is referenced inside the inner function`

A variable is defined in outer function and used in inside nested function is belong to enclosing scope

if a variable is not defined in the outer function but used in the nested function, it is not part of the enclosing scope
###***Global Scope***
variable created in global scope which is accessible in any programe within  the file

###***Built- in-Scope***
special type of scope which is created when interpreter start ,store all the keyword and built in name

## LEGB in Function Arguments

The scope of a function argument is strictly local to the function in which it is declared
This means the argument variable is created when the function is called, can only be accessed or modified within the functions body, and is destroyed immediately after the function completes execution

```python
# default argument
def my_function(nam: str,dept = "AI"):
  # name and dept are both local scope
  print(f"Name is:{nam} and department is:{dept}")

my_function("Ayat")
print(nam) # NameError: name 'names' is not defined
```

```python
# positional argument
def party(dress_code: str,colour:str):
  print(f"Girls wearing: {dress_code} with {colour} colours")

print(party("saree","black"))
print(dress_code) # NameError
print(colour)  # nameError
```
