###***Local Scope*** : variable name which we define inside the function which is only accessible inside the function

###***Enclosing Scope*** : also known as `non-local space` which only exists if we use nested functions , refer the scope of outer function that contains inner function

###***Global Scope*** : variable created in global scope which is accessible in any programe within  the file

###***Built- in-Scope*** : special type of scope which is created when interpreter start ,store all the keyword and built in name
