```python
# problem statatement:find out the value is odd and even and return True = even and False = odd
def is_even(number: int)->bool:
  if number % 2 == 0:
    return True
  else:
    return False

user_input = int(input("Enter your number:"))
is_even(user_input)
```

```python
# find largest number from 3 number given solve without max() function
def find_largest(num1: int, num2: int, num3: int):
  if num1 >= num2 and num1 >= num3:
    return num1
  elif num2 >= num1 and num2 >= num3:
    return num2
  elif num3 >= num1 and num3 >= num2:
    return num3



find_largest(10,25,7)
find_largest(-3,-8, -1)
find_largest(5,5,2)
```

```python
# grad calculator
def grade_calculator(marks: int):
  if marks >= 90 and marks <= 100:
    return 'A'
  elif marks >= 80 and marks <= 89:
    return 'B'
  elif marks >= 70 and marks <= 79:
    return 'C'
  elif marks >= 60 and marks <= 69:
    return 'D'
  elif marks < 60 and marks > 0:
    return 'F'
  elif  marks > 100 or marks <= 0 :
    return "Invalid Marks"
cal = grade_calculator(-3)
print(cal)
```

```python
# count vowels in string

def count_vowel(s: str):
  count = 0
  vowel_list = ['a','A','e','E','i','I','o','O','u','U']
  for i in s:
    if i in vowel_list:
      count+=1

  return f"Number of vowel in your string:{count}"

count_vowel("hello")
count_vowel("python Programming")
count_vowel("rythm")
count_vowel('aeiouAE')
```

```python
# sum even number from list
def sum_even_numbers(numbers: list):
  sum = 0
  for i in numbers:
    if i %2 == 0:
      sum +=i
  return sum
my_sum = sum_even_numbers([-4,-3,0,5,8])
my_sum = sum_even_numbers([])
print(my_sum)
```

```python
# remove duplicate value from list return unique value
def remove_duplicate(items: list):
  unique_list = []
  for item in items:
    if item in unique_list:
      continue
    unique_list.append(item)
  return unique_list


remove_duplicate([1,2,2,3,1,4])
remove_duplicate(['a','b','a','c'])
remove_duplicate([])
```

```python
# count word frequency from sentence in dictionary and treat uppercase/lowercase same word
def count_word_frequency(sentence: str):
  word_dic = {}
  for word in sentence.casefold().split(): # .casefold method is use for case insensitiveness
      if word in word_dic:
        word_dic[word]+=1
      else:
        word_dic[word] = 1
  return word_dic



count_word_frequency("i am am rabia  I i")
count_word_frequency("cat cat dog DOG CAT")
```

```python
# check the text is palindrome or not and return True if word is palindrom and ignore whitespace and capitilazation
def is_palindrom(text: str)-> bool:
  t = "".join(text.split())
  reverse_str = t[::-1]
  # print(t)
  # print(reverse_str)

  if t.casefold() == reverse_str.casefold():
    return True
  else:
    return False


is_palindrom('madam')
is_palindrom('python')
is_palindrom('  LeV el  ')
is_palindrom('A man a plan a canal panama')
```
