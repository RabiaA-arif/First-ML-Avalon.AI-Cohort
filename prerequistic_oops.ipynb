```python
# remove duplicate value from list return unique value
def remove_duplicate(items: list):
  unique_list = []
  for item in items:
    if item in unique_list:
      continue
    unique_list.append(item)
  return unique_list


remove_duplicate([1,2,2,3,1,4])
remove_duplicate(['a','b','a','c'])
remove_duplicate([])
```

```python
# check the text is palindrome or not and return True if word is palindrom and ignore whitespace and capitilazation
def is_palindrom(text: str)-> bool:
  t = "".join(text.split())
  reverse_str = t[::-1]
  # print(t)
  # print(reverse_str)

  if t.casefold() == reverse_str.casefold():
    return True
  else:
    return False


is_palindrom('madam')
is_palindrom('python')
is_palindrom('  LeV el  ')
is_palindrom('A man a plan a canal panama')
```

```python
import math
# function which check the prime number
def is_prime(number: int):
 if number <=1:
  return False
 if number == 2:
    return True
 if number % 2 == 0:
   return False
 for i in range(3,int(math.sqrt(number))+1 ,2):
    # print(i)
    # print(math.sqrt(number)+1)
    if number % i == 0:
      return False
 return True


is_prime(7)
is_prime(10)
is_prime(1)
is_prime(-5)
```

```python
# calculate total cart take dictionary as a cart which contain item and price
def calculate_cart_total(cart: dict, tax_rate=0):
    sub_total = []
    for i in cart:
      # print(cart[i])
      sub_total.append(cart[i])
    sum_total= sum(sub_total)
    # print(sum_total)
    tax = sum_total * tax_rate
    # print(tax)
    total = sum_total + tax
    return total

calculate_cart_total({"pen":50,"register":500})
calculate_cart_total({"book":500},0.10)
calculate_cart_total({},0.18)
```

```python
# function which take student name and list of marks and return dictionary return name ,total marks,average of marks and pass
def student_result(name: str , marks: list):
    total_marks = sum(marks)
    average_marks = round(total_marks/len(marks),2)
    # round(average_marks,2)
    status = ''
    if average_marks > 40:
      status = "pass"
    else:
      status = "Fail"
    student_dic = {}
    student_dic["Name"] = name
    student_dic["Total Marks"] = total_marks
    student_dic["Average"] = average_marks
    student_dic["Status"] = status
    return student_dic


student_result('Ali', [80, 70, 90])
student_result('Sara', [30, 45, 35])
```

## **Understanding Of Python Byte code using `dis` module and stack**

`dis` module is python built-in-standard library for disassemble python byte code into human readable format

when we run python source code python compile it into byte and than exceute using python virtual machine

dis module help us to see how our code is working in low level instruction

for apply dis module on our code our code must be in A function,A string,A object,A class format

```python
import dis

my_code = """
a = 10
b = 20
c = a + b
print(c)
"""
dis.dis(my_code)

# 2           2 LOAD_CONST               0 (10)  # pushes 10 onto the stack  ,stack:[10]
#            4 STORE_NAME               0 (a)  # pop the top value and store into a  , stack:[]
# 3           6 LOAD_CONST               1 (20) # push 20 onto the stack         , stack:[20]
#            8 STORE_NAME               1 (b)  # pop the top value and store into b ,    stack :[]
#  4          10 LOAD_NAME                0 (a) # push value on stack : [10]
#             12 LOAD_NAME                1 (b) # push value on stack : [10,20]
#             14 BINARY_OP                0 (+) # pop the value from stack and perform + and push backe to stack :[30]
#             18 STORE_NAME               2 (c)  # pop 30 and store it into c , stack:[]
# 5          20 PUSH_NULL
#            22 LOAD_NAME                3 (print) # push print onto stacl
#            24 LOAD_NAME                2 (c) # push c value on stack
#            26 CALL                     1     # call function with 1 argument :print(30)
#            34 POP_TOP                         # print statement return None so python automatically remove from stack
#            36 RETURN_CONST             2 (None)  # overall execution result is none
```

## Cpython Execution flow of Above Code


| Step | Operation                            | State or result                                   |
| ---: | ------------------------------------ | ------------------------------------------------- |
|    1 | Create the string `"java"`           | A string object exists.                           |
|    2 | Create a list containing that string | `['java']`                                        |
|    3 | Bind `lang` to the list              | `lang → ['java']`                                 |
|    4 | Call `add_items(lang)`               | `items` refers to the same list object as `lang`. |
|    5 | Execute `items.append("python")`     | The list becomes `['java', 'python']`.            |
|    6 | Reach the end of the function        | The function implicitly returns `None`.           |
|    7 | Execute `print(add_items(lang))`     | `print(None)` displays `None`.                    |
