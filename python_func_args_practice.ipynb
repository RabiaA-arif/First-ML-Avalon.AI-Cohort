## Default function arguments:

1,default arguments are the value which is provided during the function definition

2,become option during function call

3,providing value to default arguments during function call it override the value

4, python default argument evaluate only in once during the function definition not on every time we call the function

5,jub hum koi ek default value date hain tou wo use save kar lata ha and next function call main us exact value ko memory main sa utha ka daata ha

```python
def sub(a,b=50,c=21):
  return (a-b-c)

print(sub(7))
print(sub(3,5))
print(sub(2,3,6))
print(sub(a=2,c=5))
```

```python
import datetime

def is_ongoing(self,today=datetime.date.today()):
  return (today >=self.start_date) and (today <=self.end_date)

print(is_ongoing(datetime.date(2026,3,3)))

# if user run this function 7,14,2026 the default value is exact same store and we call function it return same value
# why this happen :behind the scene
# when python read code it compile function code in function object in memory
# "Any default values you assign are evaluated immediately
# and packed into a special read-only tuple called __defaults__ attached directly to that function object."
```
