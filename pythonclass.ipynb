
Exercise 2: Customer Search System (for-else concept)
A restaurant has a list of registered customers.
Task:
Explain what happens when you search for a customer who exists.
Explain what happens when you search for a customer who does NOT exist.
Mention when the system will show “not found
Understanding for-else + break logic.

```python
registered_customer = ["Rabia","Aimen","Hoorain","Ayat"]
user_input=input("Enter customer name:")
for customer in registered_customer:
  if customer == user_input:
    print("customer is found")

  elif customer is not registered_customer:
    print("customer is not found")
    break
```
